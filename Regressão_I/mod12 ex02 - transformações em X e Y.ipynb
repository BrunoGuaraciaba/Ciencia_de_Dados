# Regressão 01 - tarefa 02 - transformações em X e Y

### 1. Melhorando o ajuste no percentual de gorjetas

a. Vamos partir do modelo feito na última tarefa, o que relaciona ```tip_pct``` e ```net_bill```. Carregue a base, os pacotes e reproduza este modelo aqui.

````python
import pandas as pd
import numpy as np
import seaborn as sns
import statsmodels.formula.api as smf

# carregar base
df = sns.load_dataset('tips')

# criar variáveis necessárias
df['tip_pct'] = df['tip'] / df['total_bill']
df = df.rename(columns={'total_bill': 'net_bill'})
````

````python
modelo1 = smf.ols('tip_pct ~ net_bill', data=df).fit()
print(modelo1.summary())
````

### 2. ajuste o modelo de ```tip_pct``` em função do logaritmo de ```net_bill```.

````python
df['log_net_bill'] = np.log(df['net_bill'])

modelo2 = smf.ols('tip_pct ~ log_net_bill', data=df).fit()
print(modelo2.summary())
````

###  3. ajuste o modelo de ```tip_pct``` em função de um polinômio de segundo grau em ```net_bill```.

````python
df['net_bill_sq'] = df['net_bill']**2

modelo3 = smf.ols('tip_pct ~ net_bill + net_bill_sq', data=df).fit()
print(modelo3.summary())
````

###  4. ajuste um modelo no log do ```tip_pct``` em função de ```net_bill```. Calcule o $R^2$ também em ```tip_pct```

````python
df['log_tip_pct'] = np.log(df['tip_pct'])

modelo4 = smf.ols('log_tip_pct ~ net_bill', data=df).fit()
print(modelo4.summary())

# previsão e retorno para escala original
df['pred4'] = np.exp(modelo4.predict(df))
````

###  5. ajuste um modelo no log do ```tip_pct``` em função do log de ```net_bill```. Calcule o $R^2$ também em ```tip_pct```.

````python
modelo5 = smf.ols('log_tip_pct ~ log_net_bill', data=df).fit()
print(modelo5.summary())

# previsão e retorno para escala original
df['pred5'] = np.exp(modelo5.predict(df))
````

###  6. qual modelo resultou no melhor ajuste?

````python
print("Modelo 1 R²:", modelo1.rsquared)
print("Modelo 2 R²:", modelo2.rsquared)
print("Modelo 3 R²:", modelo3.rsquared)
print("Modelo 4 R²:", modelo4.rsquared)
print("Modelo 5 R²:", modelo5.rsquared)
````

O modelo que apresentou melhor ajuste foi o modelo com transformação log-log (log do tip_pct em função do log do net_bill), pois apresentou maior R² e melhor comportamento dos resíduos, indicando uma relação mais adequada entre as variáveis.
