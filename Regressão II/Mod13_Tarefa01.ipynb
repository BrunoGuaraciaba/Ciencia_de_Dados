```python
#df = pd.read_csv('previsao_de_renda.csv')
import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf

# carregar base
df = pd.read_csv('previsao_de_renda.csv')

# criar variável alvo (log da renda)
df['log_renda'] = np.log(df['renda'])

# visualizar colunas
df.columns
```

1. Ajuste um modelo para prever log(renda) considerando todas as covariáveis disponíveis.
    - Utilizando os recursos do Patsy, coloque as variáveis qualitativas como *dummies*.
    - Mantenha sempre a categoria mais frequente como casela de referência
    - Avalie os parâmetros e veja se parecem fazer sentido prático.  


2. Remova a variável menos significante e analise:
    - Observe os indicadores que vimos, e avalie se o modelo melhorou ou piorou na sua opinião.

3. Siga removendo as variáveis menos significantes, sempre que o *p-value* for menor que 5%. Compare o modelo final com o inicial. Observe os indicadores e conclua se o modelo parece melhor. 
    

```python
# Criar modelo de regressão múltipla
modelo = smf.ols(
    'log_renda ~ idade + tempo_emprego + qtd_filhos + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(educacao) + C(estado_civil) + C(tipo_residencia)',
    data=df
).fit()

print(modelo.summary())
```

```python
# Remover variável menos significativa
modelo2 = smf.ols(
    'log_renda ~ idade + tempo_emprego + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(educacao) + C(estado_civil) + C(tipo_residencia)',
    data=df
).fit()

print(modelo2.summary())
```
