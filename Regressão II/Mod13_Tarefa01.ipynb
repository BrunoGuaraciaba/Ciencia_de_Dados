1. Ajuste um modelo para prever log(renda) considerando todas as covariáveis disponíveis.
    - Utilizando os recursos do Patsy, coloque as variáveis qualitativas como *dummies*.
    - Mantenha sempre a categoria mais frequente como casela de referência
    - Avalie os parâmetros e veja se parecem fazer sentido prático.  


2. Remova a variável menos significante e analise:
    - Observe os indicadores que vimos, e avalie se o modelo melhorou ou piorou na sua opinião.
    - Observe os parâmetros e veja se algum se alterou muito.  


3. Siga removendo as variáveis menos significantes, sempre que o *p-value* for menor que 5%. Compare o modelo final com o inicial. Observe os indicadores e conclua se o modelo parece melhor. 
    

```python

import pandas as pd
import numpy as np
import statsmodels.api as sm
import statsmodels.formula.api as smf
from statsmodels.stats.outliers_influence import variance_inflation_factor

# carregar base
df = pd.read_csv('previsao_de_renda.csv')

# criar variável alvo (log da renda)
df['log_renda'] = np.log(df['renda'])

# visualizar colunas
df.columns
```

```python
df.info()
```

```python
# Criar modelo de regressão múltipla , Modelo completo
modelo1 = smf.ols(
    'log_renda ~ idade + tempo_emprego + qtd_filhos + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(educacao) + C(estado_civil) + C(tipo_residencia)',
    data=df
).fit()

print(modelo1.summary())
```

```python
# Remover variável menos significativa
modelo2 = smf.ols(
    'log_renda ~ idade + tempo_emprego + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(educacao) + C(estado_civil) + C(tipo_residencia)',
    data=df
).fit()

print(modelo2.summary())
```

```python
modelo3 = smf.ols(
    'log_renda ~ idade + tempo_emprego + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(educacao) + C(estado_civil)',
    data=df
).fit()

print(modelo3.summary())
```

```python

modelo4 = smf.ols(
    'log_renda ~ idade + tempo_emprego + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(estado_civil)',
    data=df
).fit()

print(modelo4.summary())
```

```python
modelo_final = smf.ols(
    'log_renda ~ tempo_emprego + '
    'C(sexo) + C(posse_de_veiculo) + C(posse_de_imovel) + '
    'C(tipo_renda) + C(estado_civil)',
    data=df
).fit()

print(modelo_final.summary())
```

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = modelo_final.model.exog

vif = pd.DataFrame()
vif["variavel"] = modelo_final.model.exog_names
vif["VIF"] = [variance_inflation_factor(X, i) for i in range(X.shape[1])]

print(vif)
```

O modelo inicial apresentou um bom ajuste, porém com diversas variáveis não significativas estatisticamente (p-value > 5%). A partir disso, foi realizado um processo iterativo de remoção das variáveis menos significativas.

Ao longo das iterações, observou-se que o R² ajustado se manteve relativamente estável, indicando que as variáveis removidas não contribuíam de forma relevante para o modelo.

Além disso, foi avaliada a multicolinearidade por meio do VIF, sendo identificada possível redundância entre variáveis como idade e tempo de emprego, o que motivou ajustes adicionais.

O modelo final apresentou menor complexidade, melhor interpretabilidade e manteve um bom nível de explicação da variável resposta, sendo considerado mais adequado.
