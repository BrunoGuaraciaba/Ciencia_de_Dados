```python
# Análise Gráfica - Base SINASC 2019

## 1. Imports necessários


import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração visual
sns.set(style='whitegrid')
plt.rcParams['figure.figsize'] = (10,6)


# 2. Carregar as bases de dados


# Março
mar = pd.read_csv('SINASC_RO_2019_MAR.csv')

# Abril
abr = pd.read_csv('SINASC_RO_2019_ABR.csv')

# Maio
mai = pd.read_csv('SINASC_RO_2019_MAI.csv')

# Junho
jun = pd.read_csv('SINASC_RO_2019_JUN.csv')

# Dezembro
dez = pd.read_csv('SINASC_RO_2019_DEZ.csv')




# 3. Adicionar coluna de mês


mar['MES'] = 'Março'
abr['MES'] = 'Abril'
mai['MES'] = 'Maio'
jun['MES'] = 'Junho'
dez['MES'] = 'Dezembro'



# 4. Unificar as bases


sinasc = pd.concat([mar, abr, mai, jun, dez], ignore_index=True)

# Visualizar informações
sinasc.info()



# 5. Quantidade de nascimentos por mês


plt.figure(figsize=(10,6))

sns.countplot(data=sinasc, x='MES', order=['Março', 'Abril', 'Maio', 'Junho', 'Dezembro'])

plt.title('Quantidade de Nascimentos por Mês')
plt.xlabel('Mês')
plt.ylabel('Quantidade')

plt.show()



# 6. Distribuição da idade das mães

plt.figure(figsize=(10,6))

sns.histplot(data=sinasc, x='IDADEMAE', bins=30, kde=True)

plt.title('Distribuição da Idade das Mães')
plt.xlabel('Idade da Mãe')
plt.ylabel('Frequência')

plt.show()


# 7. Peso do bebê ao nascer por mês

plt.figure(figsize=(10,6))

sns.boxplot(data=sinasc, x='MES', y='PESO', order=['Março', 'Abril', 'Maio', 'Junho', 'Dezembro'])

plt.title('Peso do Bebê por Mês')
plt.xlabel('Mês')
plt.ylabel('Peso')

plt.show()


# 8. Escolaridade da mãe

plt.figure(figsize=(12,6))

sns.countplot(data=sinasc, y='ESCMAE')

plt.title('Escolaridade da Mãe')
plt.xlabel('Quantidade')
plt.ylabel('Escolaridade')

plt.show()


# 9. Tipo de gravidez


plt.figure(figsize=(8,5))

sns.countplot(data=sinasc, x='GRAVIDEZ')

plt.title('Tipo de Gravidez')
plt.xlabel('Tipo')
plt.ylabel('Quantidade')

plt.show()


# 10. Sexo do bebê

plt.figure(figsize=(8,5))

sns.countplot(data=sinasc, x='SEXO')

plt.title('Sexo do Bebê')
plt.xlabel('Sexo')
plt.ylabel('Quantidade')

plt.show()


# 11. Relação entre idade da mãe e peso do bebê

plt.figure(figsize=(10,6))

sns.scatterplot(data=sinasc, x='IDADEMAE', y='PESO', alpha=0.5)

plt.title('Idade da Mãe x Peso do Bebê')
plt.xlabel('Idade da Mãe')
plt.ylabel('Peso do Bebê')

plt.show()


# 12. Resumo estatístico

sinasc[['IDADEMAE', 'PESO']].describe()

```

```python

# IDEIAS PARA AUTOMATIZAÇÃO

# 1) Criar uma lista com os arquivos automaticamente
import glob

arquivos = glob.glob('SINASC_RO_*.csv')

# 2) Ler todos os arquivos automaticamente
lista_df = []

for arquivo in arquivos:
    df_temp = pd.read_csv(arquivo)

    # Extrair nome do mês automaticamente
    nome_mes = arquivo.split('_')[-1].replace('.csv', '')

    df_temp['MES'] = nome_mes

    lista_df.append(df_temp)

# 3) Concatenar todos os arquivos
sinasc_auto = pd.concat(lista_df, ignore_index=True)

# 4) Criar função para gráficos

def grafico_countplot(dataframe, coluna, titulo):
    plt.figure(figsize=(10,6))
    sns.countplot(data=dataframe, x=coluna)
    plt.title(titulo)
    plt.show()

# Exemplo de uso
grafico_countplot(sinasc_auto, 'SEXO', 'Sexo do Bebê')


```

# Explicação

A automatização ajuda a evitar retrabalho nos próximos meses ou anos.

Com o uso do módulo glob, o código consegue localizar automaticamente todos os arquivos CSV da pasta, sem necessidade de informar manualmente cada arquivo.

Além disso, a criação de funções para geração de gráficos permite reutilizar o mesmo código para diferentes colunas e análises, deixando o notebook mais organizado, escalável e fácil de manter.

Dessa forma, para adicionar novos meses ou novos anos, basta colocar os arquivos CSV na pasta do projeto que o código fará a leitura e consolidação automaticamente.
